```python
bad_age = clean_df[(clean_df["age"] < 10) | (clean_df["age"] > 100)]

print("Number of impossible age values:", len(bad_age))
print(bad_age[["learner_id", "age"]].to_string(index=False))
```

```python
clean_df.loc[(clean_df["age"] < 10) | (clean_df["age"] > 100), "age"] = np.nan
```

```python
print("Impossible ages remaining:",
      ((clean_df["age"] < 10) | (clean_df["age"] > 100)).sum())

print("Missing ages now:", clean_df["age"].isna().sum())
```

```python
bad_hours = clean_df[clean_df["hours_per_week_pledged"] > 168]

print("Number of impossible pledged-hours values:", len(bad_hours))
print(bad_hours[["learner_id", "hours_per_week_pledged"]].to_string(index=False))
```

```python
clean_df.loc[clean_df["hours_per_week_pledged"] > 168, "hours_per_week_pledged"] = np.nan
```

```python
print(
    "Impossible pledged-hours remaining:",
    (clean_df["hours_per_week_pledged"] > 168).sum()
)

print(
    "Missing pledged-hours now:",
    clean_df["hours_per_week_pledged"].isna().sum()
)
```
